```python
cdf= df.drop(['mvmt','casem','bracem','Unnamed: 0','condition'],axis=1)
```

```python
cdf
```

```python
cdf[["price","size","yop"]].head(10)
```

```python
cdf["price"] = cdf["price"].str.replace("$", "", regex=False)#Regex is a pattern used to find, match, or manipulate specific text in a string.
cdf["price"] = cdf["price"].str.replace(",", "", regex=False)
cdf['price']=pd.to_numeric(cdf["price"], errors="coerce")
cdf['price'].head(10)
```

```python
cdf['price']=cdf['price'].fillna(cdf['price'].median())
```

```python
cdf["yop"] = cdf["yop"].str.extract(r'(\d{4})')
cdf["yop"] = pd.to_numeric(cdf["yop"], errors="coerce")
cdf['yop'].head(10)
```

```python
cdf["yop"] = cdf["yop"].fillna(cdf["yop"].median())
```

```python
cdf['size'].value_counts().head(20)
```

```python
cdf["size"] = cdf["size"].str.replace(" mm", "", regex=False)
cdf["size"] = pd.to_numeric(cdf["size"], errors="coerce")
print(cdf['size'].head(10))
print(cdf['size'].dtype)
```

```python
cdf["size"] = cdf["size"].fillna(cdf["size"].median())
```

```python
cdf[['price','yop','size']].isnull().sum()
```

```python
cdf.isnull().sum()
```

```python
cdf.select_dtypes(include="object").isnull().sum()
```

```python
cdf['brand']=cdf['brand'].fillna(cdf['brand'].mode()[0])
```

```python
cdf["model"] = cdf["model"].fillna("Unknown")
```

```python
cdf["cond"] = cdf["cond"].fillna("Unknown")
```

```python
cdf['sex']=cdf['sex'].fillna("Unknown")
```

```python
cdf.isnull().sum()
```

```python
cdf.drop('name',axis=1,inplace=True)
cdf
```

```python
plt.figure(figsize=(8,5))
plt.hist(cdf["price"], bins=50)
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

**Most watches are concentrated in the lower price range, while a few expensive luxury watches create a right-skewed distribution.**

```python
top_brands = cdf["brand"].value_counts().head(10)
plt.figure(figsize=(10,5))
plt.bar(top_brands.index, top_brands.values)
plt.title("Top 10 Watch Brands")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

**A few brands dominate the dataset, while the remaining brands have relatively fewer listings.**

```python
condition = cdf["cond"].value_counts()
plt.figure(figsize=(7,5))
plt.bar(condition.index, condition.values)
plt.title("Condition Distribution")
plt.xlabel("Condition")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

**Most watches belong to the Unknown condition category, while Very Good is the most common among the known conditions.**

```python
gender = cdf["sex"].value_counts()
plt.figure(figsize=(6,5))
plt.bar(gender.index, gender.values)
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

**Most watches in the dataset are designed for Men, followed by the Unknown category, while Women watches are comparatively fewer.**

```python
cdf[["price","yop","size"]].corr()
```

```python
corr=cdf[["price","yop","size"]].corr()
plt.figure(figsize=(6,5))
sns.heatmap(corr,annot=True, cmap="viridis",linewidths=1)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(cdf["price"],vert=False)
plt.title("Price Boxplot")
plt.ylabel("Price")
plt.show()
```

**A large number of high-priced watches appear as outliers, which is expected in a luxury watch dataset.**

```python
plt.figure(figsize=(6,5))
plt.boxplot(cdf["size"],vert=False)
plt.title("Boxplot of Watch Size")
plt.ylabel("Size (mm)")
plt.show()
```

**The size distribution contains very few outliers, indicating that most watches fall within a similar size range.**

```python
plt.figure(figsize=(6,5))
plt.boxplot(cdf["yop"],vert=False)
plt.title("Boxplot of Year of Production")
plt.ylabel("Year")
plt.show()
```

**The Year of Production shows only a few outliers, suggesting that most watches were manufactured within a relatively consistent time period.**
